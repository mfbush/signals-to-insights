## Setup and imports

All libraries the notebook uses are imported in the next cell, so a missing package shows up here rather than partway through. The cell prints nothing when it works.

````python
import sys

import numpy as np
import matplotlib.pyplot as plt
````

## Parameters

The one number this notebook depends on is the wavelength at which the standards were measured. It is set here and used in the axis label.

````python
WAVELENGTH = 520  # nm; the wavelength at which all six standards were measured
````

## 1. Load the calibration series

The file `data/uvvis_calibration.csv` has a header row and then one row per standard: concentration in µM, then absorbance in AU. The next cell reads it with a relative path, i.e., a path that starts from this notebook's folder, so it works on every laptop. The output is a small table of six rows, from 0 to 25 µM.

````python
data = np.loadtxt("data/uvvis_calibration.csv", delimiter=",", skiprows=1)
concentration = data[:, 0]  # µM
absorbance = data[:, 1]  # AU

print("concentration (µM)   absorbance (AU)")
for c, a in zip(concentration, absorbance):
    print(f"{c:18.1f}   {a:15.3f}")
````

## 2. Plot the series

The next cell plots absorbance against concentration, one marker per standard, with the units on both axes. This is the cell you change in section 3.

````python
fig, ax = plt.subplots()
ax.plot(concentration, absorbance, "o", label="standards")
ax.set_xlabel("Concentration (µM)")
ax.set_ylabel(f"Absorbance at {WAVELENGTH} nm (AU)")
ax.set_title("Session 1: UV-Vis calibration series")
ax.legend()
fig.tight_layout()
plt.show()
````

The plot shows six round markers rising from 0.002 AU at 0 µM to 1.405 AU at 25 µM.
